```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/kaggle/input/datasets/sandeepkhairwar/netflix-dataset/netflix_titles.csv")
```

```python
df.head()
```

```python
print('shape of dataset', df.shape)
print(df.info())
```

```python
print("Total Titles:", df.shape[0])

print("Movies:", len(df[df['type']=='Movie']))

print("TV Shows:", len(df[df['type']=='TV Show']))

print("Most Common Rating:", df['rating'].mode()[0])

print("Top Country:", df['country'].mode()[0])
```

```python
This script provides a quick summary of the dataset, including the total number of titles, the
breakdown of movies versus TV shows, and the most frequent rating and country.
```

```python
print(df.isnull().sum())
```

```python
df['country'] = df['country'].fillna('Unknown')
df['director'] = df['director'].fillna('Not Available')
df['cast'] = df['cast'].fillna('Not Available')
```

```python
type_count = df['type'].value_counts()

plt.figure(figsize=(5,4))
type_count.plot(kind='bar')
plt.title('Movies vs TV Shows')
plt.ylabel('Count')
plt.show()
```

```python
# This code counts the number of movies versus TV shows in the dataset and visualizes the comparison using a bar chart,
complete with a title and a y-axis label.
```

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_countries.plot(kind='bar')
plt.title('Top 10 Countries with Netflix Content')
plt.ylabel('Count')
plt.show()
```

```python
# A bar chart displaying the top 10 countries by volume of Netflix content. It highlights the
geographical distribution of the platform's library, showing which nations dominate the production of available titles
```

```python
release_trend = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(10,5))
release_trend.plot()
plt.title('Netflix Content Release Trend')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.show()
```

```python
# The line chart below illustrates the annual trend of content releases on Netflix, highlighting the
growth in the number of titles added to the platform over the years.
```

```python
top_rating = df['rating'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_rating.plot(kind='bar')
plt.title('Top Content Ratings')
plt.ylabel('Count')
plt.show()
```

```python
# A bar chart illustrating the top 10 content ratings in the dataset based on their frequency. The Y-axis represents the total count for each rating type, allowing for an easy comparison of the most common content categories
```

```python
top_genres = df['listed_in'].str.split(', ').explode().value_counts().head(10)

plt.figure(figsize=(8,5))
top_genres.plot(kind='bar')
plt.title('Top 10 Genres')
plt.ylabel('Count')
plt.show()
```

```python
# This code extracts the top 10 most frequent genres from the dataset and visualizes them using a vertical bar chart.
```
